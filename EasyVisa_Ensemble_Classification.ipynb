# **Problem Statement**


## Context


Business communities in the United States are facing high demand for human resources, but one of the constant challenges is identifying and attracting the right talent, which is perhaps the most important element in remaining competitive. Companies in the United States look for hard-working, talented, and qualified individuals both locally as well as abroad.

The Immigration and Nationality Act (INA) of the US permits foreign workers to come to the United States to work on either a temporary or permanent basis. The act also protects US workers against adverse impacts on their wages or working conditions by ensuring US employers' compliance with statutory requirements when they hire foreign workers to fill workforce shortages. The immigration programs are administered by the Office of Foreign Labor Certification (OFLC).

OFLC processes job certification applications for employers seeking to bring foreign workers into the United States and grants certifications in those cases where employers can demonstrate that there are not sufficient US workers available to perform the work at wages that meet or exceed the wage paid for the occupation in the area of intended employment.


## Objective


In FY 2016, the OFLC processed 775,979 employer applications for 1,699,957 positions for temporary and permanent labor certifications. This was a nine percent increase in the overall number of processed applications from the previous year. The process of reviewing every case is becoming a tedious task as the number of applicants is increasing every year.

The increasing number of applicants every year calls for a Machine Learning based solution that can help in shortlisting the candidates having higher chances of VISA approval. OFLC has hired the firm EasyVisa for data-driven solutions. You as a data  scientist at EasyVisa have to analyze the data provided and, with the help of a classification model:

* Facilitate the process of visa approvals.
* Recommend a suitable profile for the applicants for whom the visa should be certified or denied based on the drivers that significantly influence the case status.


## Data Description


The data contains the different attributes of employee and the employer. The detailed data dictionary is given below.

* case_id: ID of each visa application
* continent: Information of continent the employee
* education_of_employee: Information of education of the employee
* has_job_experience: Does the employee has any job experience? Y= Yes; N = No
* requires_job_training: Does the employee require any job training? Y = Yes; N = No
* no_of_employees: Number of employees in the employer's company
* yr_of_estab: Year in which the employer's company was established
* region_of_employment: Information of foreign worker's intended region of employment in the US.
* prevailing_wage:  Average wage paid to similarly employed workers in a specific occupation in the area of intended employment. The purpose of the prevailing wage is to ensure that the foreign worker is not underpaid compared to other workers offering the same or similar service in the same area of employment.
* unit_of_wage: Unit of prevailing wage. Values include Hourly, Weekly, Monthly, and Yearly.
* full_time_position: Is the position of work full-time? Y = Full Time Position; N = Part Time Position
* case_status:  Flag indicating if the Visa was certified or denied

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, Patch

fig, ax = plt.subplots(1, 1, figsize=(8, 15))

steps = [
    ("1. Load & Audit Data\n(shape, dtypes, duplicate & missing-value checks)", '#808080'),
    ("2. Clean & Engineer Features\n(fix no_of_employees sign errors,\nnormalize prevailing_wage to annual_wage)", '#808080'),
    ("3. Exploratory Data Analysis\n(univariate + bivariate analysis,\ncorrelation heatmap, Leading Questions)", '#2E86C1'),
    ("4. Preprocess Data\n(outlier check, encode target & predictors,\n60/20/20 stratified train/val/test split)", '#2E86C1'),
    ("5. Address Class Imbalance\n(Original vs. SMOTE Oversampled vs.\nRandomUnderSampler Undersampled)", '#2E86C1'),
    ("6. Baseline Model Building\n(6 models x 3 resampling variants)", '#2E7D32'),
    ("7. Select Best Variant &\nHyperparameter Tuning\n(RandomizedSearchCV on Random Forest,\nAdaBoost, Gradient Boosting, XGBoost)", '#2E7D32'),
    ("8. Compare Tuned Models &\nSelect Final Model\n(both-class check -> XGBoost selected)", '#2E7D32'),
    ("9. Final Test Evaluation &\nFeature Importance\n(test set touched once, for XGBoost only)", '#CC5500'),
    ("10. Actionable Insights &\nRecommendations", '#CC5500'),
]

n = len(steps)
box_h = 1.0
gap = 0.4
y_top = n * (box_h + gap)

for i, (label, color) in enumerate(steps):
    y = y_top - i * (box_h + gap)
    ax.add_patch(FancyBboxPatch(
        (0.5, y - box_h), 9, box_h,
        boxstyle="round,pad=0.08,rounding_size=0.12",
        facecolor=color, edgecolor='black', linewidth=1
    ))
    ax.text(5, y - box_h / 2, label, ha='center', va='center',
            fontsize=9, color='white', fontweight='bold')
    if i < n - 1:
        ax.annotate('', xy=(5, y - box_h - gap + 0.05), xytext=(5, y - box_h - 0.03),
                     arrowprops=dict(arrowstyle='-|>', color='black', linewidth=1.5))

ax.set_xlim(0, 10)
ax.set_ylim(0, y_top + 0.3)
ax.axis('off')
ax.set_title('End-to-End Process Flow', fontsize=12, fontweight='bold')

legend_elements = [
    Patch(facecolor='#808080', edgecolor='black', label='Data Understanding & Cleaning'),
    Patch(facecolor='#2E86C1', edgecolor='black', label='EDA & Preprocessing'),
    Patch(facecolor='#2E7D32', edgecolor='black', label='Modeling & Tuning'),
    Patch(facecolor='#CC5500', edgecolor='black', label='Evaluation & Insights'),
]
ax.legend(handles=legend_elements, loc='upper center', bbox_to_anchor=(0.5, -0.01),
          ncol=2, fontsize=8, frameon=False)

plt.tight_layout()
plt.show()
```

<p style="color:black"><p><b>What this shows:</b> the ten-step path this notebook follows from raw data to final recommendations, color-coded into four phases. <b>Data Understanding & Cleaning</b> (gray) loads and audits the raw dataset, then corrects the <code style="background:none; padding:0; border:none;">no_of_employees</code> sign errors and normalizes <code style="background:none; padding:0; border:none;">prevailing_wage</code> into <code style="background:none; padding:0; border:none;">annual_wage</code>. <b>EDA & Preprocessing</b> (blue) explores the cleaned data to answer the Leading Questions, then encodes and splits it 60/20/20 before building the three resampling variants (Original, Oversampled, Undersampled) compared throughout the notebook. <b>Modeling & Tuning</b> (green) fits six baseline models on each variant, selects the strongest variant, tunes the four best candidates with <code style="background:none; padding:0; border:none;">RandomizedSearchCV</code>, and picks XGBoost as the final model using a both-class performance check rather than a single headline score. <b>Evaluation & Insights</b> (orange) evaluates the final model exactly once on the untouched test set and closes with the Key Insights and Recommendations that follow from it.</p></p>

```python
# Custom palette: gray, green, orange, light blue, purple, goldenrod -- six colors so every
# categorical column in this dataset (continent has the most levels, at 6) gets its own distinct color
PALETTE = ['#808080', '#2E7D32', '#CC5500', '#2E86C1', '#7B3F99', '#B8860B']
FONTSIZE = 10

# Heatmaps (correlation matrix + confusion matrices) use a dedicated gray -> green -> blue scale
HEAT_CMAP = LinearSegmentedColormap.from_list('gray_green_blue', ['#808080', '#2E7D32', '#2E86C1'])


def category_color_map(categories):
    """
    Maps each category value to a fixed PALETTE color, keyed off the ALPHABETICAL order of
    the category values -- not the order they happen to be displayed in. This is what makes
    a given category (e.g. "Certified", "Bachelor's") the same color every time it appears
    in any bar or box plot in this notebook, regardless of chart type or display order.
    """
    cats_sorted = sorted(str(c) for c in set(categories))
    return {c: PALETTE[i % len(PALETTE)] for i, c in enumerate(cats_sorted)}


def get_bar_color(i=0):
    """Cycles through PALETTE -- used where bars represent different features/metrics rather than category levels of one column."""
    return PALETTE[i % len(PALETTE)]


def get_hist_color(i=0):
    """Cycles through PALETTE for labeled_histogram -- keeps a grid of distributions internally consistent."""
    return PALETTE[i % len(PALETTE)]


def get_box_color(i=0):
    """Cycles through PALETTE for labeled_boxplot -- keeps a grid of comparisons internally consistent."""
    return PALETTE[i % len(PALETTE)]


def style_ax(ax, title, xlabel_rotation=0):
    """
    Applies consistent formatting to any matplotlib/seaborn axis -- call this at the end
    of any plotting function so every visual in the notebook shares the same look.
    All x-axis tick labels are horizontal by default, per notebook-wide styling.

    ax: the matplotlib axis the plot was drawn on
    title: text to use as the subplot title
    xlabel_rotation: degrees to rotate x tick labels (default 0 -- horizontal -- across the whole notebook)
    """
    ax.set_title(title, fontsize=FONTSIZE)
    ax.set_xlabel('')
    ax.set_ylabel(ax.get_ylabel(), fontsize=FONTSIZE)
    ax.tick_params(axis='x', labelsize=FONTSIZE)
    ax.tick_params(axis='y', labelsize=FONTSIZE)
    # plt.setp(..., ha='center', rotation_mode='anchor') keeps each rotated label centered
    # directly under its own bar/box -- tick_params(rotation=...) alone can leave labels
    # anchored off to one side once rotated.
    plt.setp(ax.get_xticklabels(), rotation=xlabel_rotation, ha='center', rotation_mode='anchor')
    # Always show a legend when the axis actually has labeled artists (e.g. mean/median
    # lines, hue-based series) -- ax.get_legend() only re-styles an existing legend, so
    # we build one from any labeled handles first.
    handles, labels = ax.get_legend_handles_labels()
    if labels:
        ax.legend(fontsize=FONTSIZE - 2)
    ax.grid(axis='y', linestyle='--', alpha=0.7)


def labeled_barplot(data, feature, ax, perc=False, n=None):
    """
    Barplot with percentage/count labels at the top. Each bar is colored by its own
    category value (via category_color_map), so e.g. "Bachelor's" is always the same
    color everywhere education_of_employee is plotted in this notebook.

    data: dataframe
    feature: dataframe column
    ax: matplotlib axis to draw the plot on
    perc: whether to display percentages instead of count (default is False)
    n: displays the top n category levels (default is None, i.e., display all levels)
    """

    total = len(data[feature])
    order = data[feature].value_counts().index[:n]
    color_map = category_color_map(data[feature].dropna().unique())
    palette = [color_map[str(c)] for c in order]

    sns.countplot(
        data=data,
        x=feature,
        order=order,
        hue=feature,
        hue_order=order,
        palette=palette,
        legend=False,
        dodge=False,
        ax=ax,
    )

    for p in ax.patches:
        if perc == True:
            label = "{:.1f}%".format(100 * p.get_height() / total)
        else:
            label = int(p.get_height())

        x = p.get_x() + p.get_width() / 2
        y = p.get_height()

        ax.annotate(
            label, (x, y), ha="center", va="center",
            fontsize=FONTSIZE - 1, xytext=(0, 5), textcoords="offset points",
        )

    style_ax(ax, title=feature, xlabel_rotation=0)


def labeled_histogram(data, feature, ax, color=None, bins=30, kde=True, color_index=0):
    """
    Histogram with mean and median lines marked, and a legend identifying them.

    data: dataframe
    feature: dataframe column
    ax: matplotlib axis to draw the plot on
    color: bar color (overrides color_index/get_hist_color if given)
    bins: number of histogram bins (default 30)
    kde: whether to overlay a density curve (default True)
    color_index: which PALETTE slot to use via get_hist_color (default 0)
    """

    sns.histplot(
        data=data, x=feature, bins=bins, kde=kde,
        color=color if color else get_hist_color(color_index),
        ax=ax,
    )

    mean_val = data[feature].mean()
    median_val = data[feature].median()

    ax.axvline(mean_val, color='#CC5500', linestyle='--', linewidth=1.5, label=f'Mean: {mean_val:.1f}')
    ax.axvline(median_val, color='#2E7D32', linestyle='-', linewidth=1.5, label=f'Median: {median_val:.1f}')

    style_ax(ax, title=feature, xlabel_rotation=0)


def labeled_boxplot(data, feature, target, ax):
    """
    Boxplot for a numerical feature against a categorical target variable. Each box is
    colored by its own category value (via category_color_map), so e.g. "Certified" is
    always the same color everywhere case_status is plotted in this notebook.

    data: dataframe
    feature: numerical column to plot
    target: categorical column for comparison (e.g., 'case_status')
    ax: matplotlib axis to draw the plot on
    """
    order = sorted(str(c) for c in data[target].dropna().unique())
    color_map = category_color_map(data[target].dropna().unique())
    palette = [color_map[c] for c in order]

    sns.boxplot(
        data=data, y=feature, x=target, order=order,
        hue=target, hue_order=order, palette=palette, legend=False,
        dodge=False, ax=ax,
    )
    style_ax(ax, title=f'{feature} by {target}', xlabel_rotation=0)


def barplot_by_status(data, feature, target='case_status', ax=None, perc=False, n=None):
    """
    Grouped/dodged countplot of `feature`, split by `target` (e.g. case_status), colored by
    the target's own category values (via category_color_map) -- so "Certified"/"Denied"
    always show up in the same colors as everywhere else they're plotted in this notebook.
    """
    order = data[feature].value_counts().index[:n]
    hue_order = sorted(str(c) for c in data[target].dropna().unique())
    color_map = category_color_map(data[target].dropna().unique())
    hue_palette = [color_map[c] for c in hue_order]

    sns.countplot(
        data=data, x=feature, order=order,
        hue=target, hue_order=hue_order, palette=hue_palette,
        ax=ax,
    )

    total = len(data[feature])
    for p in ax.patches:
        label = "{:.1f}%".format(100 * p.get_height() / total) if perc else int(p.get_height())
        x = p.get_x() + p.get_width() / 2
        y = p.get_height()
        ax.annotate(
            label, (x, y), ha="center", va="center",
            fontsize=FONTSIZE - 1, xytext=(0, 5), textcoords="offset points",
        )

    style_ax(ax, title=feature, xlabel_rotation=0)
    ax.legend(title=target, fontsize=FONTSIZE - 2)


def model_performance_classification_sklearn(model, predictors, target):
    """
    Computes accuracy, recall, precision, and F1 (all w.r.t. the positive class, Certified=1)
    for a fitted model against the given predictors/target.
    """
    pred = model.predict(predictors)
    return pd.DataFrame({
        "Accuracy": [accuracy_score(target, pred)],
        "Recall": [recall_score(target, pred)],
        "Precision": [precision_score(target, pred)],
        "F1": [f1_score(target, pred)],
    })


def plot_confusion_matrix(model, predictors, target, ax, title):
    """
    Confusion matrix heatmap with count + percent-of-total annotations, using the
    notebook-wide gray/green/blue colormap and horizontal x tick labels.
    """
    cm = confusion_matrix(target, model.predict(predictors))
    labels = np.asarray([
        f"{count:d}\n{count / cm.sum():.2%}" for count in cm.flatten()
    ]).reshape(2, 2)
    sns.heatmap(
        cm, annot=labels, fmt='', cmap=HEAT_CMAP,
        xticklabels=['Denied', 'Certified'], yticklabels=['Denied', 'Certified'],
        cbar_kws={'label': 'Count'}, ax=ax,
    )
    ax.set_xlabel('Predicted', fontsize=FONTSIZE)
    ax.set_ylabel('Actual', fontsize=FONTSIZE)
    ax.set_title(title, fontsize=FONTSIZE)
    ax.tick_params(axis='x', labelsize=FONTSIZE)
    ax.tick_params(axis='y', rotation=0, labelsize=FONTSIZE)
    plt.setp(ax.get_xticklabels(), rotation=0, ha='center', rotation_mode='anchor')
```

#### Fixing the negative values in the number of employees column

<p style="color:black"><b>What was done:</b> 33 rows have a negative no_of_employees value. These are treated as data-entry sign errors (a company cannot have a negative headcount) and corrected in place by taking the absolute value, rather than dropped -- dropping them would throw away otherwise-usable rows over a one-column formatting issue.</p>

```python
# Taking the absolute value for number of employees -- treated as a data-entry sign error
data['no_of_employees'] = data['no_of_employees'].abs()
```

#### Normalizing prevailing_wage to an annual figure


<p style="color:black"><b>Why this is needed:</b> <code style="background:none; padding:0; border:none;">prevailing_wage</code> mixes Hourly/Weekly/Monthly/Yearly amounts in the same column (see <code style="background:none; padding:0; border:none;">unit_of_wage</code>), so a $25/hour wage and a $25,000/year wage currently look directly comparable to any chart or model that reads this column raw. The cell below normalizes everything to an annual figure -- using a 40-hour work week for full-time positions and 35 hours/week for part-time -- before dropping the now-redundant raw column. <code style="background:none; padding:0; border:none;">unit_of_wage</code> itself is kept as its own categorical feature, since it carries signal beyond wage level (hourly-paid applicants are certified far less often regardless of pay).</p>

```python
HOURS_FULL_TIME = 40
HOURS_PART_TIME = 35
WEEKS_PER_YEAR = 52
wage_mult_base = {'Week': 52, 'Month': 12, 'Year': 1}

unit = data['unit_of_wage'].astype(str)
is_hourly = unit == 'Hour'
hours_per_week = np.where(data['full_time_position'].astype(str) == 'N', HOURS_PART_TIME, HOURS_FULL_TIME)

annual_wage = data['prevailing_wage'] * unit.map(wage_mult_base)
data['annual_wage'] = annual_wage.where(~is_hourly, data['prevailing_wage'] * hours_per_week * WEEKS_PER_YEAR)
data.drop(['prevailing_wage'], axis=1, inplace=True)

data[['annual_wage']].describe()
```

### Outlier Check


```python
numeric_cols = data.select_dtypes(include=np.number).columns.tolist()

outlier_summary = []
for col in numeric_cols:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    n_outliers = ((data[col] < lower_bound) | (data[col] > upper_bound)).sum()
    outlier_summary.append({
        'feature': col, 'Q1': round(Q1, 2), 'Q3': round(Q3, 2),
        'lower_bound': round(lower_bound, 2), 'upper_bound': round(upper_bound, 2),
        'n_outliers': n_outliers, 'pct_outliers': round(100 * n_outliers / len(data), 2),
    })

outlier_summary = pd.DataFrame(outlier_summary).sort_values('pct_outliers', ascending=False)
display(outlier_summary)

fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

feature_order = outlier_summary['feature'].tolist()
feature_color_map = category_color_map(feature_order)
bar_colors = [feature_color_map[f] for f in feature_order]

bars = axes[0].bar(feature_order, outlier_summary['pct_outliers'], color=bar_colors)
for bar, pct in zip(bars, outlier_summary['pct_outliers']):
    axes[0].annotate(f'{pct:.2f}%', (bar.get_x() + bar.get_width() / 2, bar.get_height()),
                      ha='center', va='bottom', fontsize=FONTSIZE - 1, xytext=(0, 3), textcoords='offset points')
axes[0].margins(y=0.15)
axes[0].set_ylabel('% Outliers (IQR method)', fontsize=FONTSIZE)
style_ax(axes[0], title='Outlier Percentage by Feature', xlabel_rotation=0)

melted = data[numeric_cols].melt(var_name='feature', value_name='value')
box_colors = [feature_color_map[f] for f in numeric_cols]
sns.boxplot(data=melted, x='feature', y='value', hue='feature', hue_order=numeric_cols,
            palette=box_colors, legend=False, dodge=False, ax=axes[1])
style_ax(axes[1], title='Distribution of Numerical Features', xlabel_rotation=0)

plt.tight_layout(pad=2.0)
plt.show()
```

<p style="color:black"><b>Outlier treatment decision:</b> The extreme values flagged above (e.g., a 600k-employee company, wages above $300k) are genuine data points, not data-entry errors -- there is no evidence they are corrupted (unlike the earlier no_of_employees sign issue, which was treated as an error and corrected). Every model built in this notebook is a tree-based ensemble (Decision Tree, Bagging, Random Forest, AdaBoost, Gradient Boosting, XGBoost), which splits on thresholds rather than relying on distance, so these outliers are left untreated rather than capped or removed -- doing so would discard real signal without meaningfully changing how these models learn.</p>

### Data Preparation for modeling


<p style="color:black"><b>What was done:</b> The target is encoded as <code style="background:none; padding:0; border:none;">case_status</code> = 1 for Certified and 0 for Denied -- this notebook is framed around identifying which applicants are likely to be <b>certified</b>. Categorical predictors are one-hot encoded, and the data is split 60/20/20 into train, validation, and test sets (stratified on the target) so that the test set can be touched exactly once, at the very end, for the single chosen model.</p>

```python
# Certified = 1: frames the model around identifying which applicants are likely to be certified
data['case_status'] = data['case_status'].apply(lambda x: 1 if x == 'Certified' else 0)
data['case_status'].value_counts(normalize=True)
```

```python
predictor_cat_cols = [c for c in categorical_cols if c != 'case_status']
X = pd.get_dummies(data.drop('case_status', axis=1), columns=predictor_cat_cols, drop_first=True)
y = data['case_status']

X.head()
```

```python
# Splitting data into training, validation and test set (60/20/20):
# first split into a temporary set and a test set...
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.20, random_state=1, stratify=y
)

# ...then split the temporary set into train and validation
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.25, random_state=1, stratify=y_temp
)

print("X_train shape:", X_train.shape)
print("X_val shape:", X_val.shape)
print("X_test shape:", X_test.shape)
print("y_train distribution:\n", y_train.value_counts(normalize=True))
print("y_val distribution:\n", y_val.value_counts(normalize=True))
print("y_test distribution:\n", y_test.value_counts(normalize=True))
```

```python
sm = SMOTE(random_state=1, sampling_strategy=1)
X_train_over, y_train_over = sm.fit_resample(X_train, y_train)

rus = RandomUnderSampler(random_state=1, sampling_strategy=1)
X_train_un, y_train_un = rus.fit_resample(X_train, y_train)

print("Original training shape:", X_train.shape, y_train.value_counts().to_dict())
print("Oversampled training shape:", X_train_over.shape, y_train_over.value_counts().to_dict())
print("Undersampled training shape:", X_train_un.shape, y_train_un.value_counts().to_dict())
```

# **Model Building**


## Model evaluation criterion


<p style="color:black"><b>Model can make wrong predictions as:</b><br>
1. Model predicts an application will be Certified, but it should have been Denied.<br>
2. Model predicts an application will be Denied, but it should have been Certified.<br><br>
<b>Which case is more important?</b> Both carry real cost. If a genuinely denial-bound application is predicted Certified, a position goes to a worker despite the employer not having demonstrated the labor shortage the INA requires. If a genuinely certifiable application is predicted Denied, a legitimate employer loses a qualified worker and the U.S. economy loses that worker's contribution.<br><br>
<b>How to reduce the losses?</b> F1 score is used as the primary comparison metric, computed by default on Certified (the positive class, since the target is framed around identifying likely certifications). Because Certified is also the majority class (66.8% of applications), a model can look strong on this metric while doing a mediocre job at the arguably more useful task of flagging applications headed for denial -- so a dedicated Both-class performance check (via classification_report) is added later in this notebook to surface Denied-side Recall/Precision/F1 for every tuned model as well. Class imbalance is additionally handled through resampling (oversampling/undersampling, compared below) and class_weight="balanced" is used on the baseline Random Forest as a further imbalance-aware comparison point.</p>

## Model Building - Original Data


```python
models = {
    "Decision Tree": DecisionTreeClassifier(random_state=1),
    "Bagging": BaggingClassifier(random_state=1),
    "Random Forest": RandomForestClassifier(random_state=1, class_weight="balanced"),
    "AdaBoost": AdaBoostClassifier(random_state=1),
    "Gradient Boosting": GradientBoostingClassifier(random_state=1),
    "XGBoost": XGBClassifier(random_state=1, eval_metric="logloss"),
}

results = []
for name, model in models.items():
    model.fit(X_train, y_train)
    train_perf = model_performance_classification_sklearn(model, X_train, y_train)
    val_perf = model_performance_classification_sklearn(model, X_val, y_val)
    results.append({
        "Model": name,
        "Train F1": train_perf["F1"].iloc[0], "Validation F1": val_perf["F1"].iloc[0],
        "Train-Validation F1 Gap": train_perf["F1"].iloc[0] - val_perf["F1"].iloc[0],
        "Validation Recall": val_perf["Recall"].iloc[0], "Validation Precision": val_perf["Precision"].iloc[0],
    })

results_original = pd.DataFrame(results).sort_values("Validation F1", ascending=False)
results_original
```

<p style="color:black"><b>Comment on model performance (Original Data):</b> Gradient Boosting has the best validation F1 (0.824), with AdaBoost close behind (0.818) and XGBoost not far off (0.814) -- all three have small train-validation gaps (0.004, 0.001, and 0.079 respectively). Random Forest (0.801), Bagging (0.777), and Decision Tree (0.750) trail, and Random Forest, Bagging, and Decision Tree all show much larger train-validation gaps (0.20-0.25), meaning they overfit the training set's specific patterns far more than Gradient Boosting or AdaBoost do.</p>

## Model Building - Oversampled Data


```python
models = {
    "Decision Tree": DecisionTreeClassifier(random_state=1),
    "Bagging": BaggingClassifier(random_state=1),
    "Random Forest": RandomForestClassifier(random_state=1, class_weight="balanced"),
    "AdaBoost": AdaBoostClassifier(random_state=1),
    "Gradient Boosting": GradientBoostingClassifier(random_state=1),
    "XGBoost": XGBClassifier(random_state=1, eval_metric="logloss"),
}

results = []
for name, model in models.items():
    model.fit(X_train_over, y_train_over)
    train_perf = model_performance_classification_sklearn(model, X_train_over, y_train_over)
    val_perf = model_performance_classification_sklearn(model, X_val, y_val)
    results.append({
        "Model": name,
        "Train F1": train_perf["F1"].iloc[0], "Validation F1": val_perf["F1"].iloc[0],
        "Train-Validation F1 Gap": train_perf["F1"].iloc[0] - val_perf["F1"].iloc[0],
        "Validation Recall": val_perf["Recall"].iloc[0], "Validation Precision": val_perf["Precision"].iloc[0],
    })

results_oversampled = pd.DataFrame(results).sort_values("Validation F1", ascending=False)
results_oversampled
```

Random Forest (0.795), Bagging (0.757), and Decision Tree (0.736) again trail and overfit heavily. Based on all three sections, the Original (non-resampled) data produced the strongest and most consistent validation results overall.</p>

# **Model Performance Improvement**


## **Note**


<p style="color:black">Hyperparameter search below uses <code style="background:none; padding:0; border:none;">RandomizedSearchCV</code> (a randomized search over each grid, rather than exhaustively evaluating every combination) so that runtime stays reasonable while still covering each parameter space well. ## A. Additional Methodology Context


<p style="color:black"><p><b>End-to-end pipeline:</b> raw data &rarr; sign-error correction on <code style="background:none; padding:0; border:none;">no_of_employees</code> &rarr; <code style="background:none; padding:0; border:none;">prevailing_wage</code> normalized to <code style="background:none; padding:0; border:none;">annual_wage</code> &rarr; categorical encoding (one-hot) &rarr; stratified 60/20/20 train/validation/test split &rarr; six baseline models compared across three training-data variants (Original, Oversampled, Undersampled) &rarr; the four strongest, most tunable candidates (Random Forest, AdaBoost, Gradient Boosting, XGBoost) re-trained with <code style="background:none; padding:0; border:none;">RandomizedSearchCV</code> on the best-performing variant &rarr; both-class performance check &rarr; single final model chosen and evaluated once on the untouched test set.</p>
<p><b>Why tree-based ensembles only:</b> the feature set is dominated by categorical variables and a few heavily skewed numeric ones. Tree-based models split on thresholds rather than distances, so they don't require feature scaling and are naturally robust to the outliers and skew documented in the EDA section, which is why no linear or distance-based model (e.g., logistic regression, KNN) was included in the comparison.</p>
<p><b>Why <code style="background:none; padding:0; border:none;">RandomizedSearchCV</code> over <code style="background:none; padding:0; border:none;">GridSearchCV</code>:</b> each parameter grid below has well over 100 possible combinations once every hyperparameter is combined; with 5-fold cross-validation, an exhaustive <code style="background:none; padding:0; border:none;">GridSearchCV</code> would multiply that out to a runtime that isn't practical here. <code style="background:none; padding:0; border:none;">RandomizedSearchCV</code> samples a fixed number of random combinations (<code style="background:none; padding:0; border:none;">n_iter</code> = 12-15 depending on the model) from the same grid, giving broad coverage of the space in a fraction of the time -- the tradeoff is that it is not guaranteed to find the single best combination the way an exhaustive search would.</p>
<p><b>Why F1 with a both-class check, rather than Accuracy alone:</b> Accuracy on a 66.8% / 33.2% imbalanced target rewards a model for defaulting to the majority (Certified) class. F1 on the Certified class was used as the primary tuning/comparison metric per the framing of the target, but every tuned model is additionally re-scored with <code style="background:none; padding:0; border:none;">classification_report</code> on both classes so that Denied-side performance -- arguably the more operationally useful signal -- is never hidden behind a strong headline number.</p>
<p><b>Reproducibility:</b> <code style="background:none; padding:0; border:none;">random_state=1</code> is fixed on every model, split, and resampler in the notebook, so re-running all cells in order reproduces the exact numbers reported throughout.</p></p>

## C. Assumptions Made


<p style="color:black"><p><b>Negative <code style="background:none; padding:0; border:none;">no_of_employees</code> values are data-entry sign errors, not a separate meaningful category.</b> These 33 rows were corrected via absolute value rather than dropped or flagged as a new feature.</p>
<p><b>Hourly-wage-to-annual conversion assumes standard work-week lengths.</b> Part-time (<code style="background:none; padding:0; border:none;">full_time_position == 'N'</code>) hourly applicants are assumed to work 35 hours/week and full-time applicants 40 hours/week, both across 52 weeks/year, to make <code style="background:none; padding:0; border:none;">annual_wage</code> comparable across all four <code style="background:none; padding:0; border:none;">unit_of_wage</code> categories. Only 19 of 2,157 hourly-paid applicants are part-time, so this assumption has minimal effect on the overall results.</p>
<p><b>Outliers in the numeric features are genuine, not data-entry errors, and are left untreated.</b> This assumption depends on every model being tree-based (threshold-splitting rather than distance-based); it would need to be revisited if a linear or distance-based model were introduced later.</p>
<p><b>The dataset is assumed complete and clean after the initial audit</b> -- 0 missing values and 0 duplicate rows were confirmed programmatically, so no imputation or de-duplication step was needed.</p>
<p><b>Certified is treated as the positive class (label 1)</b> throughout, framing every model around identifying applicants likely to be certified, per this project's explicit direction.</p>
<p><b>Categorical features are one-hot encoded without imposing ordinality</b> -- for example, <code style="background:none; padding:0; border:none;">education_of_employee</code>'s natural High School &lt; Bachelor's &lt; Master's &lt; Doctorate ordering is not encoded numerically. Tree-based models can recover ordinal-like structure through repeated splits, so this is assumed not to meaningfully cost performance relative to ordinal encoding.</p>
<p><b>The 60/20/20 stratified split is assumed representative going forward</b> -- no time-based (out-of-time) split was used, since the dataset has no application-date field to test for drift over time.</p>
<p><b>Class imbalance is assumed to be adequately addressed through resampling and <code style="background:none; padding:0; border:none;">class_weight="balanced"</code></b> (on the baseline Random Forest only) rather than through threshold-moving or a custom cost-sensitive loss function.</p>
<p><b>The <code style="background:none; padding:0; border:none;">RandomizedSearchCV</code> parameter grids are assumed to sufficiently cover each model's useful tuning space</b> within a practical runtime budget; a larger grid or an exhaustive <code style="background:none; padding:0; border:none;">GridSearchCV</code> could plausibly find marginally better configurations at meaningfully higher computational cost.</p></p>

## D. Limitations and Future Work


<p style="color:black"><p><b>Denied-class recall tops out around 50% even for the final model.</b> XGBoost still predicts Certified for roughly half of the applications that will actually be denied -- this model should support a human reviewer's triage process, not replace it or auto-approve applications.</p>
<p><b>No temporal validation was possible.</b> The dataset has no application-date field, so it is unknown whether the relationships found here (e.g., education and job-experience effects) hold steady if OFLC's policies, the applicant pool, or the labor market shift over time.</p>
<p><b>The feature set does not capture qualitative case detail</b> that a human OFLC officer might weigh -- specific occupation/job duties, the exact prevailing-wage benchmark source, or employer certification history. Incorporating occupation-code-level detail, if available, is a natural extension.</p>
<p><b>Cost-sensitive learning was not explored.</b> Class imbalance was addressed via resampling and <code style="background:none; padding:0; border:none;">class_weight="balanced"</code> only; explicitly weighting a false-Certified prediction differently from a false-Denied prediction in the loss function (reflecting the asymmetric real-world costs discussed in the Model Evaluation Criterion) is a reasonable next step.</p>
<p><b>Classification-threshold tuning was not explored.</b> Every model above uses the default 0.5 probability threshold; adjusting it on the final XGBoost model is a low-cost way to trade some Certified precision for improved Denied recall, if that trade is judged worthwhile downstream.</p></p>

## E. Glossary of Evaluation Metrics


<p style="color:black"><p><b>Accuracy:</b> the share of all predictions (both classes combined) that were correct.</p>
<p><b>Precision (for a class):</b> of everything the model predicted as that class, the share that actually was that class.</p>
<p><b>Recall (for a class):</b> of everything that actually was that class, the share the model correctly identified.</p>
<p><b>F1 score:</b> the harmonic mean of precision and recall for a class -- it balances both rather than favoring one at the other's expense, which is why it is used as the primary metric throughout this notebook.</p>
<p><b>Confusion matrix:</b> a 2x2 table of True Negative / False Positive / False Negative / True Positive counts (here, "positive" means Certified) that every metric above is computed from.</p></p>
